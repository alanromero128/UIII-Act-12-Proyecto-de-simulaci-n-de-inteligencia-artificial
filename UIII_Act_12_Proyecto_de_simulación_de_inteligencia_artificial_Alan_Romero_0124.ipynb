### Clase Arbitro

```python
class Arbitro:
    def __init__(self, fed, partidos, nivel, deporte, sanciones, nac):
        self.federacion = fed
        self.partidos_dirigidos = partidos
        self.nivel_certificacion = nivel
        self.deporte = deporte
        self.sanciones_aplicadas = sanciones
        self.nacionalidad = nac
    def agregar_arbitro(lista):
        print("\n--- Registro de Nuevo Árbitro ---")
        fed = input("Federación: ")
        partidos = int(input("Partidos dirigidos: "))
        nivel = input("Nivel de certificación: ")
        deporte = input("Deporte: ")
        sanciones = int(input("Sanciones aplicadas: "))
        nac = input("Nacionalidad: ")

        nuevo = Arbitro(fed, partidos, nivel, deporte, sanciones, nac)
        lista.append(nuevo)
        print("¡Árbitro agregado con éxito!")
    def listar_arbitros(lista):
        if not lista:
            print("\nNo hay árbitros registrados.")
            return
        print("\n" + "="*120)
        print(f"{'#':<3} | {'Deporte':<15} | {'Federación':<15} | {'Certificación':<18} | {'Partidos':<10} | {'Sanciones':<10} | {'Nacionalidad':<15}")
        print("-" * 120)
        for i, a in enumerate(lista):
            print(f"{i:<3} | {a.deporte:<15} | {a.federacion:<15} | {a.nivel_certificacion:<18} | {a.partidos_dirigidos:<10} | {a.sanciones_aplicadas:<10} | {a.nacionalidad:<15}")
        print("="*120)

    def actualizar_arbitro(lista):
        Arbitro.listar_arbitros(lista)
        if not lista:
            return
        idx = int(input("\nIngresa el número del árbitro a actualizar: "))

        if 0 <= idx < len(lista):
            a = lista[idx]
            print(f"\nEditando a árbitro de {a.deporte}. (Presiona Enter para mantener el valor actual)")

            fed = input(f"Nueva Federación [{a.federacion}]: ")
            if fed: a.federacion = fed

            partidos = input(f"Partidos dirigidos [{a.partidos_dirigidos}]: ")
            if partidos: a.partidos_dirigidos = int(partidos)

            nivel = input(f"Nivel certificación [{a.nivel_certificacion}]: ")
            if nivel: a.nivel_certificacion = nivel

            print("¡Datos actualizados!")
        else:
            print("Índice no válido.")

    def borrar_arbitro(lista):
        Arbitro.listar_arbitros(lista)
        if not lista:
            return
        idx = int(input("\nIngresa el número del árbitro a eliminar: "))

        if 0 <= idx < len(lista):
            lista.pop(idx)
            print("Árbitro eliminado del sistema.")
        else:
            print("Índice no válido.")


arbitros_db = []

while True:
    print("\n" + "•"*10 + " SISTEMA DE GESTIÓN DE ÁRBITROS " + "•"*10)
    print("1. Registrar Árbitro")
    print("2. Listar Árbitros")
    print("3. Actualizar Árbitro")
    print("4. Borrar Árbitro")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        Arbitro.agregar_arbitro(arbitros_db)
    elif opcion == "2":
        Arbitro.listar_arbitros(arbitros_db)
    elif opcion == "3":
        Arbitro.actualizar_arbitro(arbitros_db)
    elif opcion == "4":
        Arbitro.borrar_arbitro(arbitros_db)
    elif opcion == "5":
        print("Saliendo del sistema...")
        break
    else:
        print("Opción inválida.")
```
